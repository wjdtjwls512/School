```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
x_data = [2, 3, 4, 5]
y_data = [1, 3, 5, 7]

plt.ylim(0, 8)
plt.plot(x_data, y_data, label = 'data')

plt.title('Graph Title')
plt.xlabel('x_name')
plt.ylabel('y_name')
plt.legend()

plt.show()
```

```python
data = pd.read_excel('dust_pm25.xlsx', index_col = 'area')
plt.figure(figsize = (9, 3))

for year in range(2019, 2022):
    y_data = data[year]
    plt.plot(y_data, marker = 's', label = year)
plt.xlabel('area')
plt.ylabel('micrometer')
plt.xticks(rotation = 45)
plt.title('2019~2021 Dust Line Graph')
plt.legend()
plt.show()
```

```python
data2017 = data[2017]
plt.hist(data2017, bins = 9, label = 'pm2.5')
plt.xlabel('pm2.5')
plt.xlabel('frequency')
plt.title('2017 Dust Histogram')
plt.legend()
plt.grid()
plt.show()
```
