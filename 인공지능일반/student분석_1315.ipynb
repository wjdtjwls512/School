```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("student.csv")
df.head()
```

```python
# 공부시간의 분포

plt.hist(df['StudyHours'])
plt.xlabel('Study Hours')
plt.ylabel('Frequency')
plt.title('Study Hours Distribution')
plt.show()
```

```python
# 공부시간과 과제점수의 관계

plt.scatter(df['StudyHours'],df['AssignmentScore'])
plt.xlabel('Study Hours')
plt.ylabel('Assignment Score')
plt.title('Study Hours vs Assignment Score')
plt.show()
```

```python
# 통과와 미통과 집단별 평균 구한 후, 공부시간 비교

result_mean = df.groupby('Result')['StudyHours'].mean()
```

```python
# 두 집단을 막대그래프로 나타내기

plt.bar(result_mean.index, result_mean.values)
plt.xlabel('Result')
plt.xticks([0, 1])
plt.ylabel('Average Study Hours')
plt.title('Average Study Hours by Result')
```

```python
# 출석률과 과제점수의 관계

plt.scatter(df['Attendance'],df['AssignmentScore'])
plt.xlabel('Attendance')
plt.ylabel('Assignment Score')
plt.title('Attendance vs Assignment Score')
plt.show()
```

```python
# 수면시간과 과제점수의 관계

plt.scatter(df['Attendance'],df['StudyHours'])
plt.xlabel('Attendance')
plt.ylabel('StudyHours')
plt.title('Attendance vs StudyHours')
plt.show()
```

```python
plt.hist(df['SleepHours'])
plt.xlabel('Sleep Hours')
plt.ylabel('Frequency')
plt.title('Sleep Hours Distribution')
plt.show()
```
