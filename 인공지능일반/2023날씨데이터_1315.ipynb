```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
weather = pd.read_excel('weather2023.xlsx', index_col = 'date')
weather.head()
```

```python
print(weather.isnull().sum())
weather.drop(['rain', 'location', 'loc_name'], axis = 1, inplace = True)
weather.head()
```

```python
weather.info()
```

```python
weather.describe()
```

```python
weather.corr()
```

```python
weather.hist(bins = 30)
plt.show()
```

```python
plt.scatter(weather['cloud'], weather['sunshine'], color = 'green', alpha = 0.5) 
plt.title('cloud-sunshine scatter')
plt.xlabel('cloud')
plt.ylabel('sunshine')
plt.show()
```

```python
plt.scatter(weather['cloud'], weather['humi']) 
plt.title('cloud-humi scatter')
plt.xlabel('cloud')
plt.ylabel('humi')
plt.show()
```
