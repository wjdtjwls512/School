```python
#2
nums=list(map(int, input().split()))
cnt_1=0
cnt_2=0
for i in nums:
    if i%2==0:
        cnt_1+=i
    else:
        cnt_2+=i
print('짝수 합:',cnt_1)
print('홀수 합:',cnt_2)
```

```python
#6
pw=input()

has_alpha=False
has_digit=False

for i in pw:
    if i.isalpha():
        has_alpha=True
    if i.isdigit():
        has_digit=True

if len(pw)>=8 and has_alpha and has_digit:
    print('사용가능')
else:
    print('사용 불가능')
```

```python
#7
nums=list(map(int, input().split()))
max_nums=nums[0]
min_nums=nums[0]

for i in nums:
    if i>max_nums:
        max_nums=i
    if i<min_nums:
        min_nums=i

print('최댓값:',max_nums)
print('최솟값:',min_nums)
```

```python
#8
n=int(input())
total=0
avg=0
safe=0
for i in range(n):
    score=int(input())
    total+=score
    if score>=80:
        safe+=1
avg=total/n
print('평균:',avg)
print('합격자 수:',safe)
```

```python
#11
f=map(str, input().split())
cnt_1=0
cnt_2=0
cnt_3=0
for i in f:
    if i.endswith('py'):
        cnt_1+=1
    elif i.endswith('txt'):
        cnt_2+=1
    elif i.endswith('jpg'):
        cnt_3+=1
print('py',cnt_1)
print('txt',cnt_2)
print('jpg',cnt_3)
```
