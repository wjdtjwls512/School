```python
e=input()
e=e.lower()
vow=0
con=0
for i in e:
    if i in "aeiou":
        vow+=1
    elif i in "bcdfghjklmnpqrstvwxyz":
        con+=1
print(f"모음: {vow}개 자음: {con}개")
```
