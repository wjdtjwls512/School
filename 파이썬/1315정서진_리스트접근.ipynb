```python
#5-1
score=[60,70,75,85,90]
total=0
for i in range(len(score)):
    total+=score[i]
    avg=total//len(score)
print(avg)
for i in range(len(score)):
    if score[i]>=avg:
        print(score[i],end=' ')
```

```python
#5-2
score=[60,70,75,85,90]
total=0
for i in score:
    total+=i
    avg=total//len(score)
print(avg)
for i in score:
    if i>=avg:
        print(i,end=' ')
```
