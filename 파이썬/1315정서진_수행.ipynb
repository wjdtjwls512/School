```python
import random as r
while True:
    s=r.randint(1,10)
    e=r.randint(1,10)
    if s==e:
        print(f'{s}~{e}={s}')
        break
    if s>e:
        s,e=e,s

    total=0
    for i in range(s,e+1):
        total+=i
    print(f'{s}~{e}=>{total}')
```

```python
list1 = ["홍길동", "일지매"]

# 1. 추가: 이름을 입력받아 리스트 끝에 붙임
list1.append(input("추가할 학생: "))
print(list1)

# 2. 삭제: 이름을 입력받아 리스트에서 제거
list1.remove(input("삭제할 학생: "))
print(list1)


# 3. 변경: 바꿀 대상의 위치를 찾아서 새 이름으로 덮어쓰기
old = input("바꿀 대상: ")
new = input("새 이름: ")
list1[list1.index(old)] = new

print(f"최종 명단: {list1}")
```

```python
pw=input()
has_alpha=False
has_digit=False

for i in pw:
    if i.isalpha():
        has_alpha=True
    if i.isdigit():
        has_digit=True

if len(pw)>=8 and has_alpha and has_digit:
    print('가능')
else:
    print('불가능')
```
