```python
num=1
total=0
cnt=0
while True:
    total+=num
    cnt+=1
    num+=1
    if num>10:
        break
print(total)
print(total/cnt)
```

```python
hap=0
cnt=0
while True:
    score=input("점수:")
    if score=="q":
        print("입력 종료")
        break
    cnt+=1
    hap+=float(score)
if cnt>0:
    print("합계:",hap)
    print("평균:",hap/cnt)
else:
    print("입력된 점수 없음")
```

```python
start,end=map(int,(input().split(" ")))
total=0
if start>end:
    start,end=end,start
for i in range(start,end+1):
    total+=i
print(f"{start}~{end}까지의 합은 {total}")
```

```python
n=int(input())
for i in range(2,n+1):
    prime=True
    for j in range(2,i):
        if i%j==0:
            prime=False
            break
    if prime:
        print(i,end=" ")
```

```python
hap=0
cnt=0
while True:
    score=input("종료:문자")
    if not score.replace('.','',1).isdigit():
        print("입력 종료")
        break
    cnt+=1
    hap+=float(score)
    avg=hap/cnt
if cnt>0:
    print("합계:",hap)
    print("평균:",avg)
else:
    print("입력된 수가 없음")
```

```python
a=int(input())
b=int(a/2)+1
for i in range(1,b+1):
    for j in range(b-i):
        print(" ",end=" ")
    for j in range(i*2-1):
        print("*",end=" ")
    print()
for i in range(b+1,a+1):
    for j in range(i-b):
        print(" ",end=" ")
    for j in range((2*b-i)*2-1):
        print("*",end=" ")
    print()
```

```python
import random, time
w=['cat','dog','fox','monkey','mouse','panda','frog','snake','wolf']
n=0
i=0
print('타자게임:준비되면 엔터')
input()
start=time.time()
while i<5:
    q=random.choice(w)
    print(q)
    ans=input()
    if q==ans:
        print("pass")
        n+=1
    else:
        print("fail")
    i+=1
end=time.time()
re=end-start
print(f"pass{n}개, 시간제한10초:걸린시간:{re:4f}초")
if re>10:
    print("시간초과")
elif n<5:
    print("오답있음")
else:
    print("성공")
```
