```python
def fac(n):
    if n<=1:
        return 1
    return n*fac(n-1)

print(fac(5))
```

```python
def fibo(n):
    if n==0:
        return 0
    if n==1:
        return 1
    return fibo(n-1)+fibo(n-2)

print(fibo(0))
```

```python
def hanoi(n,a,b,c):
    if n==1:
        print(f'원반 {n}을 {a}에서 {b}로 이동')
        return
    hanoi(n-1,a,c,b)
    print(f'원반 {n}을 {a}에서 {b}로 이동')
    hanoi(n-1,c,b,a)
hanoi(3,'A','C','B')
```
