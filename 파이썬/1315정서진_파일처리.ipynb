```python
f=open('testx.csv','r',encoding='utf=8')
rows=(f.read()).split('\n')
f.close()
print(rows)
for i in rows:
    value=i.split(',')
    for j in value:
        print(j, end=' ')
    print()
```

```python
f=open('testx2.csv','r',encoding='utf=8')
rows=(f.read()).split('\n')
f.close()
print(rows)
for i in rows:
    value=i.split(',')
    total=0
    for j in value:
        total+=int(j)
    print(total, end=' ')
    print()
```
