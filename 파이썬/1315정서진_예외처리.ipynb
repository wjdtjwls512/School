```python
while True:
    try:
        n=int(input())
    except ValueError:
        print('다시 입력')
    else:
        print(n)
        break
```

```python
def lst_create():
    lst=list(map(int, input().split()))
    print('생성된 리스트',lst)
    return lst

def lst_append(lst_):
    while True:
        n=input()
        try:
            lst_.append(float(n))
        except ValueError:
            break
    print(lst_)
    return lst_

def lst_cal(lst_):
    total=0
    for i in lst_:
        total+=i
    avg=total/len(lst_)
    return total,avg

def lst_print(lst_,result):
    print('점수',lst_)
    print('평균',result)
lst=lst_create()
lst=lst_append(lst)
result=lst_cal(lst)
lst_print(lst, result)
```
