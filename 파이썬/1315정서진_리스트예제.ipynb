```python
#12
nums=[1,2,2,3,3,3,4]
d = {}
for n in nums:
    if n in d:
        d[n] += 1
    else:
        d[n] = 1
mx_cnt = 0
res = None
for k, v in d.items():
    if v > mx_cnt:
        mx_cnt = v
        res = k
print(res)
```

```python
#15
nums = [5, 2, 8, 1, 3]
n = len(nums)
for i in range(n):
    for j in range(i+1,n):
        if nums[i]>nums[j]:
            nums[i],nums[j]=nums[j],nums[i]
print(nums)
```
