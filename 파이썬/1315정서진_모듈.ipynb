```python
import math
lst=list(range(1,11))
print(lst)

mean=sum(lst)/len(lst)
print(f'평균은 {mean}')

vsum=0
for i in lst:
    vsum=vsum+(i-mean)**2
var=vsum/len(lst)
print(f'분산:{var}')

std=math.sqrt(var)
print(f'표준편차는:{std}')

gcd=math.gcd(*lst)
lcm=math.lcm(*lst)
print(f'최대공약수:{gcd} 최소공베수:{lcm}')
```
