```python
def f(n):
    k=0
    while k<n:
        k+=1
        print(k)
f(3)
```

```python
def g(n):
    k=n
    while k!=0:
        print(k)
        k-=1
g(5)
```

```python
def g(n):
    k=1
    for i in range(1,n+1):
        k*=i
    return k
g(5)
```

```python
s=0
def f(n):
    global s
    k=0
    while k<n:
        k+=1
        s+=k
f(3)
print(s)
```

```python
s=1
def g(n):
    global s
    for i in range(1,n+1):
        s*=i
g(5)
print(s)
```

```python
def g(n):
    total=0
    for i in range(1,n+1):
        if i%2==0:
            total+=i
    return total
g(10)
```
