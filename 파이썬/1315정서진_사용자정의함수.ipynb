```python
#1
def lst_create():
    lst=list(map(int, input().split()))
    print(f'생선된 리스트:{lst}')
    return lst

def lst_append(lst):
    while True:
        data=input()
        if data=='q':
            print(lst)
            return lst
        lst.append(int(data))

def lst_cal(lst):
    total=sum(lst)
    # total=0
    # for i in lst:
    #     total+=i
    avg=total/len(lst)
    return total, avg

def lst_print(lst,result):
    print(f'점수:{lst}')
    print(f'합계: {result[0]} 평균: {result[1]}')

lst=lst_create()
lst=lst_append(lst)
result=lst_cal(lst)
print(result)
lst_print(lst,result)
```

```python
class Monster:
    animal='동물'
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def say(self):
        print(f'나는 {self.name}, {self.age}살 임')
shark = Monster('상어', 7)
wolf = Monster('늑대', 3)

shark.say()
wolf.say()
print(Monster.animal)
```

```python
#부모: 기본 붕어빵
class FishBread:
    kind='DGSW붕어빵'
    def __init__(self, content):
        self.content = content
    def cook(self):
        print('붕어빵을 굽습니다')
    def eat(self):
        print(f'{self.content}붕어빵을 맛있게 먹습니다!')

#자식: 아이스크릶 붕어빵(상속)
class IceFishBread(FishBread):
    def freeze(self):
        print('아이스크림을 넣습니다')
    def eat(self):
        print(f'{self.content}붕어빵에 아이스크림을 널어 차갑게 먹습니다')

bread=IceFishBread('슈크림')

print(FishBread.kind)
bread.cook()
bread.freeze()
bread.eat()

fish=FishBread('딸기쨈')
fish.eat()
# fish.freeze()
```
