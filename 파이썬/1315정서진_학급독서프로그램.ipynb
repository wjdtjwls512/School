```python
#1
stu = int(input("학생 수: "))

dic = {}
total = 0
max_read = 0
mvp = {}
for i in range(stu):
    n, b = input().split()
    s = int(b)
    dic[n] = s
    total += s
    if s > max_read:
        max_read = s
        mvp = {n: s}
avg = total / stu
print(f'평균 독서 권수: {avg}')
avg_over = []
for k in dic:
    if dic[k] >= avg:
        avg_over.append(k)
print(f'평균 이상 학생: {avg_over}')
for k, v in mvp.items():
    print(f'최다 독서 학생: {k} {v}')
```

```python
#2
n=int(input())
a=input().split()
m=int(input())
s=input().split()
d={}
u={}
un=list()
for x in a:
  if x in s:
    d[x]=True
  else:
    d[x]=False

for x in a:
  if x not in s:
    u[x]=True
    un.append(x)

sc=list(d.values()).count(True)
uc=list(d.values()).count(False)

print(f"제출 여부: {d}")
print(f"미제출 학생: {un}")
print(f"제출수: {sc}")
print(f"미제출수: {uc}")
```
