```python
# %%
#3 사용자 정의함수 작성 – 여러 개의 값 반환
def arith(a, b):
    add = a + b
    sub = a - b

    return add, sub
i, j = arith(10, 1)
print(i)
print(j)
```

```python
# %%
#4 사용자 정의함수 작성 - 가변 매개변수 이용

#*하나
def add(*paras):
    print(paras)
    total = 0
    for para in paras:
        total += para
    return total

print(add(10))
print(add(10, 100))
print(add(10, 100, 1000))

print("---------------") #출력 구분용 실제 코드와 상관 없음

#**둘 -> 출력시 딕셔너리 형태로 받음
def student(**info):
    print(info)
student(name = "조준호")
student(name = "조준호", age = 17)
student(name = "조준호", age = 17, major = "Ai")
```

```python
# %%
#5 사용자 정의함수 작성 – 리스트 점수처리 프로그램

def list_avg(s_list):
    sum = 0
    for i in s_list:
        sum += i
    result = sum / len(s_list)
    print('점수의 평균:',result)

score = [90, 80, 70, 50, 60]
list_avg(score)
```

```python
# %%
#6 재귀함수 - 피보나치 수열
def fibo(n):
    if n <= 1:
        return n
    return fibo(n-1)+fibo(n-2)
print(fibo(10))
```

```python
# %%
#7 클래스 상속 - 알림시스템 설계
class Notification:
    platform = "공통 알람 시스템"
    method = "알림"

    def __init__(self, user):
        self.user = user

    def send(self, message):
        print(f'[{self.method}] {self.user}에게 알림: "{message}"')

    def get_user(self):
        return self.user


class EmailNotification(Notification):
    method = "이메일"


class SMSNotification(Notification):
    method = "문자"


class AppNotification(Notification):
    method = "앱푸쉬"


# 인스턴스 생성
email = EmailNotification("철수")
sms = SMSNotification("영희")
app = AppNotification("민수")

# 클래스변수 출력
print(Notification.platform)
print(EmailNotification.method)
print(SMSNotification.method)
print(AppNotification.method)

# 속성값 출력
print(email.user)
print(sms.user)
print(app.user)

# 메소드 호출
email.send("회원가입 완료")
sms.send("인증번호 1234")
app.send("새 메시지가 도착했습니다")
```

```python
# %%
#8 클래스 정의 및 사용 – 합격 여부 판별 프로그램
class Score:
    def __init__(self,score):
        self.score = score

    def is_pass(self):
        if  self.score >= 60:
            print("합격")
        else:
            print("불합격")

num1 = Score(60)
num2 = Score(40)

num1.is_pass()
num2.is_pass()
```

```python
# %%
#9 클래스 정의 및 사용 - 도서할인 프로그램
class Book:
    category = "도서"

    def __init__(self,title,price):
        self.title = title
        self.price = price
    
    def discount(self,rate):
        self.price = self.price - self.price * rate
    
    def show(self):
        print(f"{self.title}/{self.price}/{Book.category}")

B1 = Book("파이썬 입문",20000)
B1.discount(0.1)
B1.show()
```
