```python
#1
n=int(input())
scores=[]
for i in range(n):
    s=int(input())
    scores.append(s)

total=0
high=scores[0]
low=scores[0]

for s in scores:
    total+=s
    if s>high:
        high=s
    if s<low:
        low=s
avg=total/len(scores)
print('총점:',total)
print('평균:',avg)
print('최고:',high)
print('최저:',low)
```

```python
#6
n=int(input())
score={}
for i in range(n):
    name,s=input().split()
    if int(s)>=80:
        s='합격'
    else:
        s='불합격'
    score[name]=s
    print(name,score[name])
```

```python
#9
s=input()
cnt_1=0
cnt_2=0
cnt_3=0
cnt_4=0

for i in s:
    if i in 'qwertyuioplkjhgfdsazxcvbnm':
        cnt_1+=1
    elif i in '1234567890':
        cnt_2+=1
    elif i in ' ':
        cnt_3+=1
    else:
        cnt_4+=1
print(cnt_1,cnt_2,cnt_3,cnt_4)
```

```python
#11
text=input()
seen=set()
dup=set()
for ch in text:
    if ch in seen:
        dup.add(ch)
    else:
        seen.add(ch)
print(dup)
```
