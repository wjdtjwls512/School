```python
i=0
sum=0
while True:
    score=int(input())
    if score==-1:
        break
    if 0<=score<=100:
        sum+=score
        i+=1
    else:
        print("잘못된 점수입니다")
if i== 0:
    print("입력된 점수가 없습니다")
else:
    print(f"평균:{sum/i}")
```

```python
has_exclamation=False
has_number=False
while True:
    pw=input()

    for ch in pw:
        if ch=="!":
            has_exclamation=True
        if ch.isdigit():
            has_number=True
    if len(pw)>=8 and has_exclamation and has_number:
        print("사용 가능 비밀번호")
        break
    else:
        print("다시 입력")
```

```python
point=0
for i in range(1,201):
    if i%28==0:
        continue
    elif i%4==0:
        point+=4
    elif i%7==0:
        point+=7
print(f"{i}일차 총 에코 포인트:{point}")
```

```python
num = 5
squ = 0
for i in range(num):
    for j in range(i+1):
        squ += 10 ** j
    print(' '*(num-i-1)+str(squ*squ))
    squ = 0
```
