```python
import random as r
select=["가위","바위","보"]
computer=r.choice(select)

me=input("가위 바위 보중 하나를 선택: ")

print("나는",me,"를 선택")
print("컴퓨터는",computer,"를 선택")

if me=="가위" and computer=="보" or me=="바위" and computer=="가위" or me=="보" and computer=="바위":
    print("당신이 이김")

elif me=="가위" and computer=="바위" or me=="바위" and computer=="보" or me=="보" and computer=="가위":
    print("컴퓨터가 이김")

elif me==computer:
    print("비겼습니다")

else:
    print("잘못된 입력입니다")
```

```python
import random as r
computer1=r.randint(1,6)
computer2=r.randint(1,6)
list1=[computer1,computer2]
user=input("숫자 2개 입력:")
user=list(map(int,user.split(' ')))
print("com:",list1)
print("user:",user)

count=0
for i in user:
    if i in list1:
        count+=1

if count==2:
    print("1등")
elif count==1:
    print("2등")
else:
    print("3등")
```

```python
from datetime import datetime
a=input()
a=list(map(int, a.split(",")))
print(a)

if a[1]==1 or a[1]==3:
    print("남성")
elif a[1]==2 or a[1]==4:
    print("여성")

year1=a[0]//10000

if a[1]==1 or a[1]==2:
    year2=1900+year1
else:
    year2=2000+year1

today=datetime.now()

age=today.year-year2
print(f"현재나이 {age}살")
```
