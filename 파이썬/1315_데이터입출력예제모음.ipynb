```python
price=input("빵의 단가 입력:")
price=int(price)
count=int(input("빵의 개수 입력:"))
total=price*count
total=total-(total/count)
print("전체 빵값:",total)
```

```python
#BMI=몸무게/(키*키)*10000
height=input("키 입력:")
weight=input("몸무게 입력:")
height=int(height)
weight=int(weight)
bmi=weight/(height*height)*10000
print("체질량 지수=",bmi)
```

```python
pen=700
era=500
boo=2500
total=(pen*p)+(e*era)+(boo*b)
p=int(input("연필의 판매 개수는?"))
e=int(input("지우개의 판매 개수는?"))
b=int(input("책의 판매 개수는?"))
print("-"*30)
print("파이썬 문방구의 오늘 판매 내역")
print("-"*30)
print("연필:",p,"개,",p*pen,"원")
print("지우개:",e,"개,",e*era,"원")
print("공책:",b,"개,",b*boo,"원")
print("-"*30)
print("오늘 총 판매 금액 :",total,"원")
print("-"*30)
```
